```python
# Example number
n = 180*3

# Factorize the number
factors = n.factor()

# Print the factors
print(factors)  # Output: 2^2 * 3 * 5

# Extract and print the prime factors and their exponents
prime_factors = [(p, e) for p, e in factors]
print(prime_factors)  # Output: [(2, 2), (3, 1), (5, 1)]

# Extract just the prime factors
primes = [p for p, e in factors]
print(primes)  # Output: [2, 3, 5]
```
